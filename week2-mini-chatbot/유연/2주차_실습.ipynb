```python
from langchain_text_splitters import CharacterTextSplitter

char_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=250,
    chunk_overlap=50,
    length_function=len,
    is_separator_regex=False,
)

char_docs = char_splitter.create_documents([file])

print(f"[CharacterTextSplitter] 청크 개수: {len(char_docs)}")
print(char_docs[0].page_content)
print("=" * 50)
print(char_docs[1].page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=250,
    chunk_overlap=50,
    length_function=len,
    is_separator_regex=False,
)

recursive_docs = recursive_splitter.create_documents([file])

print(f"[RecursiveCharacterTextSplitter] 청크 개수: {len(recursive_docs)}")
print(recursive_docs[0].page_content)
print("=" * 50)
print(recursive_docs[1].page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_community.vectorstores import FAISS
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_core.prompts import PromptTemplate
#from langchain_openai import ChatOpenAI, OpenAIEmbeddings
```

```python
# 1. 구글 드라이브 마운트
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. PDF 파일 경로 지정 (구글 드라이브 내 실제 경로에 맞게 수정)
from langchain_community.document_loaders import PyMuPDFLoader

file_path = "/content/drive/MyDrive/tave/2주차/SPRI_AI_Brief_2023년12월호_F.pdf"
```

```python
# 단계 1: 문서 로드(Load Documents)
loader = PyMuPDFLoader(file_path)
docs = loader.load()
print(f"문서의 페이지수: {len(docs)}")
```

```python
# 단계 2: 문서 분할(Split Documents)
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=50)
split_documents = text_splitter.split_documents(docs)
print(f"분할된 청크의수: {len(split_documents)}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

# 단계 3: 임베딩(Embedding) 생성
embeddings = HuggingFaceEmbeddings(model_name="jhgan/ko-sroberta-multitask")
```

```python
# 단계 4: DB 생성(Create DB) 및 저장
# 벡터스토어를 생성합니다.
vectorstore = FAISS.from_documents(documents=split_documents, embedding=embeddings)
```

```python
# 단계 5: 검색기(Retriever) 생성
# 문서에 포함되어 있는 정보를 검색하고 생성합니다.
retriever = vectorstore.as_retriever()
```

```python
# 검색기에 쿼리를 날려 검색된 chunk 결과를 확인합니다.
retriever.invoke("삼성전자가 자체 개발한 AI 의 이름은?")
```

```python
# 단계 6: 프롬프트 생성(Create Prompt)
# 프롬프트를 생성합니다.
prompt = PromptTemplate.from_template(
    """You are an assistant for question-answering tasks.
Use the following pieces of retrieved context to answer the question.
If you don't know the answer, just say that you don't know.
Answer in Korean.

#Question:
{question}
#Context:
{context}

#Answer:"""
)
```

```python
from langchain_huggingface import HuggingFacePipeline, ChatHuggingFace

# 단계 7: 언어모델(LLM) 생성
llm_pipeline = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen2.5-3B-Instruct",
    task="text-generation",
    device=0,
    pipeline_kwargs={"max_new_tokens": 256, "temperature": 0.1, "return_full_text": False},
)
llm = ChatHuggingFace(llm=llm_pipeline)
```

```python
# 단계 8: 체인(Chain) 생성
chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
# 체인 실행(Run Chain)
# 문서에 대한 질의를 입력하고, 답변을 출력합니다.
question = "삼성전자가 자체 개발한 AI 의 이름은?"
response = chain.invoke(question)
print(response)
```
