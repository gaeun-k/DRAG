```python
import warnings, os
warnings.filterwarnings("ignore")
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="jhgan/ko-sroberta-multitask")
```

```python
text = "임베딩 테스트를 하기 위한 샘플 문장입니다."

query_result = embeddings.embed_query(text)

print("벡터 차원 수:", len(query_result))
print("앞 5개 값   :", query_result[:5])
```

```python
doc_result = embeddings.embed_documents([text, text, text, text])

print("문서 개수:", len(doc_result))
print("각 벡터 차원:", len(doc_result[0]))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sentence1 = "안녕하세요? 반갑습니다."
sentence2 = "안녕하세요? 반갑습니다!"
sentence3 = "안녕하세요? 만나서 반가워요."
sentence4 = "Hi, nice to meet you."
sentence5 = "I like to eat apples."

sentences = [sentence1, sentence2, sentence3, sentence4, sentence5]
embedded_sentences = embeddings.embed_documents(sentences)

def similarity(a, b):
    return cosine_similarity([a], [b])[0][0]

for i, sentence in enumerate(embedded_sentences):
    for j, other_sentence in enumerate(embedded_sentences):
        if i < j:
            print(f"[유사도 {similarity(sentence, other_sentence):.4f}] {sentences[i]} \t <=====> \t {sentences[j]}")
```

```python
import warnings, os
warnings.filterwarnings("ignore")
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name="jhgan/ko-sroberta-multitask")
```

```python
import os

os.makedirs("./data", exist_ok=True)

sample_text = """RAG는 검색 증강 생성을 의미하며, 질문에 답하기 전에 관련 문서를 먼저 찾아와 근거로 활용하는 방식이다.
임베딩은 텍스트를 숫자 벡터로 변환하는 과정으로, 의미가 비슷한 문장은 벡터 공간에서 가까운 위치에 놓인다.
코사인 유사도는 두 벡터가 이루는 각도를 기준으로 유사도를 측정하는 방법이다.
벡터DB는 임베딩 벡터를 저장하고 빠르게 검색할 수 있도록 돕는 저장소이다.
FAISS는 Meta가 공개한 벡터 검색 라이브러리이며 서버 없이 동작한다.
BM25는 단어의 빈도와 희귀도를 기반으로 문서의 관련성을 점수화하는 검색 알고리즘이다."""

with open("./data/appendix-keywords.txt", "w", encoding="utf-8") as f:
    f.write(sample_text)
```

```python
from langchain_classic.storage import LocalFileStore
from langchain_classic.embeddings import CacheBackedEmbeddings
from langchain_community.vectorstores import FAISS

store = LocalFileStore("./cache/")

cached_embedder = CacheBackedEmbeddings.from_bytes_store(
    underlying_embeddings=embedding,
    document_embedding_cache=store,
    namespace=embedding.model_name,   # OpenAI는 .model, 허깅페이스는 .model_name
)

# 캐시가 비어있는지 확인
print("캐시 키:", list(store.yield_keys()))
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter

raw_documents = TextLoader("./data/appendix-keywords.txt", encoding="utf-8").load()

text_splitter = CharacterTextSplitter(separator="\n", chunk_size=100, chunk_overlap=0)
documents = text_splitter.split_documents(raw_documents)

print("청크 개수:", len(documents))
```

```python
%time db = FAISS.from_documents(documents, cached_embedder)
```

```python
# 캐시된 임베딩 개수 확인 — 문서 청크만큼 늘어나 있어야 함
print("캐시된 항목 수:", len(list(store.yield_keys())))
```

```python
import warnings
warnings.filterwarnings("ignore")

from langchain_huggingface import HuggingFaceEmbeddings

model_name = "intfloat/multilingual-e5-large-instruct"   # 다국어 지원, 1024차원

hf_embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs={"device": "cpu"},          # mps → cpu (콜랩은 Mac이 아니라서)
    encode_kwargs={"normalize_embeddings": True},
)
```

```python
texts = [
    "안녕, 만나서 반가워.",
    "LangChain simplifies the process of building applications with large language models",
    "랭체인 한국어 튜토리얼은 LangChain의 공식 문서를 바탕으로 구성되어 있습니다.",
    "LangChain은 초거대 언어모델로 애플리케이션을 구축하는 과정을 단순화합니다.",
    "Retrieval-Augmented Generation (RAG) is an effective technique for improving AI responses.",
]

embedded_documents = hf_embeddings.embed_documents(texts)
embedded_query = hf_embeddings.embed_query("LangChain 에 대해서 알려주세요.")

print("차원 수:", len(embedded_query))
```

```python
import numpy as np

# 질문 하나를 문서 여러 개와 한꺼번에 비교해서 점수를 매긴다
scores = np.array(embedded_query) @ np.array(embedded_documents).T
sorted_idx = scores.argsort()[::-1]

print("[Query] LangChain 에 대해서 알려주세요.\n" + "=" * 40)
for rank, idx in enumerate(sorted_idx):
    print(f"{rank+1}위 (점수 {scores[idx]:.3f}) {texts[idx]}")
```

```python
texts_prefixed = ["passage: " + t for t in texts]   # 문서는 passage:
query_prefixed = "query: " + "LangChain 에 대해서 알려주세요."   # 질문은 query:

embedded_documents = hf_embeddings.embed_documents(texts_prefixed)
embedded_query = hf_embeddings.embed_query(query_prefixed)

scores = np.array(embedded_query) @ np.array(embedded_documents).T
sorted_idx = scores.argsort()[::-1]

print("[Query] LangChain 에 대해서 알려주세요.\n" + "=" * 40)
for rank, idx in enumerate(sorted_idx):
    print(f"{rank+1}위 (점수 {scores[idx]:.3f}) {texts[idx]}")
```
