```python
import warnings, os
warnings.filterwarnings("ignore")
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="jhgan/ko-sroberta-multitask")
```

```python
import os

os.makedirs("data", exist_ok=True)

nlp_keywords = """Word2Vec: 단어를 벡터 공간에 매핑해 의미적 유사성을 수치로 나타내는 자연어 처리 기법이다. 비슷한 맥락에서 쓰이는 단어일수록 벡터가 가깝게 배치된다.

TF-IDF: 단어의 빈도와 여러 문서에 걸친 희소성을 함께 고려해 어떤 단어가 특정 문서에서 얼마나 중요한지 나타내는 통계 수치다. 정보 검색과 텍스트 마이닝에 자주 쓰인다.

BERT: 문장 전체의 양방향 문맥을 동시에 학습하는 사전학습 언어모델이다. 다양한 자연어 처리 작업에 미세조정하여 활용할 수 있다.

토큰화: 텍스트를 문장, 단어, 서브워드 등 더 작은 단위로 나누는 과정이다. 자연어 처리 파이프라인의 가장 첫 단계에 해당한다.

임베딩: 텍스트, 이미지 등 데이터를 저차원 벡터 공간으로 변환하는 과정이다. 변환된 벡터는 의미적 유사도를 계산하는 데 사용된다."""

finance_keywords = """ESG: 환경, 사회, 지배구조를 뜻하며 기업의 지속가능성과 사회적 영향을 평가하는 비재무적 기준이다. 최근 투자 의사결정에서 중요도가 커지고 있다.

배당수익률: 주가 대비 연간 배당금의 비율을 나타내는 지표다. 배당수익률이 높을수록 주가 대비 주주환원이 크다는 뜻이다.

PER: 주가를 주당순이익으로 나눈 값으로, 기업의 이익 대비 주가 수준을 가늠하는 대표적인 밸류에이션 지표다.

변동성: 자산 가격이 일정 기간 동안 움직이는 정도를 나타내는 지표다. 변동성이 클수록 투자 위험도 커진다고 본다.

포트폴리오 분산: 여러 자산에 나누어 투자해 특정 자산의 위험이 전체 수익률에 미치는 영향을 줄이는 전략이다."""

with open("data/nlp-keywords.txt", "w", encoding="utf-8") as f:
    f.write(nlp_keywords)
with open("data/finance-keywords.txt", "w", encoding="utf-8") as f:
    f.write(finance_keywords)
```

```python
from pinecone import Pinecone, ServerlessSpec

pc = Pinecone(api_key=os.environ["PINECONE_API_KEY"])

index_name = "rag-study-index"
dimension_size = len(embeddings.embed_query("hello world"))   # 768

if index_name not in [idx["name"] for idx in pc.list_indexes()]:
    pc.create_index(
        name=index_name,
        dimension=dimension_size,
        metric="dotproduct",   # 하이브리드(Dense+Sparse) 쓰려면 dotproduct 필수
        spec=ServerlessSpec(cloud="aws", region="us-east-1"),
    )

pc_index = pc.Index(index_name)
print(pc_index.describe_index_stats())
```

```python
!pip install -q -U kiwipiepy pinecone-text

from kiwipiepy import Kiwi
from pinecone_text.sparse import BM25Encoder

kiwi = Kiwi()
KEEP_TAGS = {"NNG","NNP","NNB","NR","NP","VV","VA","MAG","XR","SL","SH","SN"}

def korean_tokenize(text):
    return [tok.form for tok in kiwi.tokenize(text) if tok.tag in KEEP_TAGS]

# 우리가 이미 만든 nlp+finance 문서로 학습
contents = [d.page_content for d in split_doc1 + split_doc2]

bm25 = BM25Encoder()
bm25.fit([" ".join(korean_tokenize(c)) for c in contents])
```

```python
from tqdm.auto import tqdm

dense_vecs = embeddings.embed_documents(contents)

vectors = []
for i, (content, dense) in enumerate(zip(contents, dense_vecs)):
    sparse = bm25.encode_documents(" ".join(korean_tokenize(content)))
    vectors.append({
        "id": f"doc-{i}",
        "values": dense,
        "sparse_values": sparse,
        "metadata": {"context": content},
    })

pc_index.upsert(vectors=vectors, namespace="study-ns")
print(pc_index.describe_index_stats())
```

```python
def hybrid_search(query, alpha=0.5, top_k=3):
    dense_q = embeddings.embed_query(query)
    sparse_q = bm25.encode_queries(" ".join(korean_tokenize(query)))

    # alpha로 두 벡터의 가중치를 조절 (RRF와 다른 방식!)
    dense_scaled = [v * alpha for v in dense_q]
    sparse_scaled = {
        "indices": sparse_q["indices"],
        "values": [v * (1 - alpha) for v in sparse_q["values"]],
    }

    result = pc_index.query(
        namespace="study-ns",
        top_k=top_k,
        vector=dense_scaled,
        sparse_vector=sparse_scaled,
        include_metadata=True,
    )
    for match in result["matches"]:
        print(f"점수 {match['score']:.3f} | {match['metadata']['context'][:40]}...")

print("=== alpha=1 (Dense만) ===")
hybrid_search("TF IDF 에 대하여 알려줘", alpha=1.0)

print("\n=== alpha=0 (Sparse만) ===")
hybrid_search("TF IDF 에 대하여 알려줘", alpha=0.0)

print("\n=== alpha=0.5 (반반) ===")
hybrid_search("TF IDF 에 대하여 알려줘", alpha=0.5)
```
